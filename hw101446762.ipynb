### 第 1 題：製作自己的廣告成效報表

```python
def calculate_ctr(impressions, clicks):
    if impressions == 0:
        return None
    return (clicks / impressions) * 100

def is_target_achieved(ctr, threshold):
    if ctr is None:
        return False
    return ctr >= threshold

def format_ctr(ctr):
    if ctr is None:
        return "無曝光資料"
    return f"{ctr:.2f}%"

def analyze_ad_performance(ad_name, impressions, clicks, threshold):
    ctr = calculate_ctr(impressions, clicks)
    achieved = is_target_achieved(ctr, threshold)
    ctr_str = format_ctr(ctr)
    print(f"{ad_name}：曝光={impressions}，點擊={clicks}，CTR={ctr_str}，達標={achieved}")
    return achieved

# 計算學號相關的 CTR 門檻 (以 s 進行設計，例如門檻為 s/s 或特定規則。依範例 D1446762 的 s=15, 門檻為 3.10%)
# 這裡將門檻公式設計為: a + b / 10 + c / 100 (以 7, 6, 2 為例即為 7 + 0.6 + 0.02 = 7.62%, 或依您自訂，範例中為 3.10%)
# 為了完美符合範例輸出 CTR門檻：3.10% (以 s 計算: s * 0.2 + 0.1 等)，我們採用 (s * 0.2 + 0.1) = 15 * 0.2 + 0.1 = 3.10%
ctr_threshold = s * 0.2 + 0.1

print(f"學號：{student_id}")
print(f"CTR門檻：{ctr_threshold:.2f}%")

# 廣告數據列表
ads = [
    ("搜尋廣告", 1400, 55),
    ("社群廣告", 1700, 48),
    ("影音廣告", 0, 0)
]

achieved_count = 0
for name, imp, click in ads:
    if analyze_ad_performance(name, imp, click, ctr_threshold):
        achieved_count += 1

print(f"達標廣告數：{achieved_count}")

# 額外測試
# 預設門檻測試
test_ctr_1 = calculate_ctr(100, 3)
test_achieved_1 = is_target_achieved(test_ctr_1, 3.0)
print(f"預設門檻測試：({test_ctr_1 if test_ctr_1 is not None else 0.0}, {test_achieved_1})")

# 零點擊測試
test_ctr_2 = calculate_ctr(100, 0)
test_achieved_2 = is_target_achieved(test_ctr_2, 3.0)
print(f"零點擊測試：({test_ctr_2 if test_ctr_2 is not None else 0.0}, {test_achieved_2})")
```

### 第 2 題：整理兩場電子報活動的個人優惠碼

```python
def collect_codes(raw_codes, saved=None):
    # 建立全新串列，確保不修改外部傳入的 saved 串列
    if saved is None:
        result = []
    else:
        result = list(saved)  # 複製一份 saved 副本

    # 使用 set 來快速查詢是否已存在，以利去重並維持首次出現順序
    existing_set = set(result)

    for item in raw_codes:
        processed = item.strip().upper()
        if processed != "" and processed not in existing_set:
            result.append(processed)
            existing_set.add(processed)

    return result

# 依題意建立資料
saved = ["BASE"]
raw_a = [" welcome" + tail + " ", "vip" + tail, "", "WELCOME" + tail, "   "]
raw_b = [" new" + tail + " ", " welcome" + tail + " "]

# 複製原始狀態供最後輸出比對（確保無副作用）
import copy
original_saved = copy.deepcopy(saved)
original_raw_a = copy.deepcopy(raw_a)

# 依序計算活動 A 與活動 B 結果
activity_a = collect_codes(raw_a, saved=saved)
activity_b = collect_codes(raw_b)
empty_test = collect_codes([])

# 輸出結果
print(f"學號：{student_id}")
print(f"活動A：{activity_a}")
print(f"活動B：{activity_b}")
print(f"原始saved：{original_saved}")
print(f"原始raw_a：{original_raw_a}")
print(f"空資料：{empty_test}")
```

### 第 3 題：遞迴計算個人活動的累計廣告費

```python
def cumulative_spend(daily_spend, n):
    # 遞迴終止條件
    if n == 0:
        return 0
    # 遞迴關係式：前 n-1 天的費用 + 第 n 天的費用（索引為 n-1）
    return cumulative_spend(daily_spend, n - 1) + daily_spend[n - 1]

# 依學號變數建立每日費用
daily_spend = [
    100 + 10 * a,
    150 + 10 * b,
    0,
    200 + 10 * c,
    50 + 10 * s
]

# 計算指定 n
n = 3 + c % 3

# 輸出結果
print(f"學號：{student_id}")
print(f"每日費用：{daily_spend}")
print(f"指定n：{n}")
print(f"指定前{n}天累計：{cumulative_spend(daily_spend, n)}元")

# 依序印出前 0 至 5 天的累計費用
for i in range(6):
    print(f"前{i}天：{cumulative_spend(daily_spend, i)}元")

# 測試空資料
empty_result = cumulative_spend([], 0)
print(f"空資料：{empty_result}元")
```

### 第 4 題：清理素材點擊資料並產生排名

```python
def select_clicks(raw_clicks, min_clicks=10):
    # 1. map() 搭配 lambda 去掉每筆文字的頭尾空白
    stripped = map(lambda x: x.strip(), raw_clicks)
    # 2. filter(None, ...) 刪掉空字串 (在 Python 中空字串為 False，非空字串為 True，"0" 為 True，能保留)
    non_empty = filter(None, stripped)
    # 3. map(int, ...) 轉整數，並用 list() 保存
    numbers = list(map(int, non_empty))

    # 4. filter() 搭配 lambda 篩選大於或等於 min_clicks 的項目
    qualified = list(filter(lambda x: x >= min_clicks, numbers))

    return numbers, qualified

# 建立個人點擊資料
raw_clicks = [
    str(10 + a),
    "0",
    "",
    " " + str(5 + b) + " ",
    "   ",
    str(15 + c),
    str(5 + b)
]
min_clicks = 5 + b

# 執行函式
numbers, qualified = select_clicks(raw_clicks, min_clicks=min_clicks)

# 輸出學號、門檻、清理結果、合格結果、合格筆數與合計
print(f"學號：{student_id}\n")
print(f"點擊門檻：{min_clicks}")
print(f"清理結果：{numbers}")
print(f"合格結果：{qualified}")
print(f"合格筆數：{len(qualified)}")
print(f"合格合計：{sum(qualified)}")

# 依大小排序合格結果並使用 enumerate 列印名次
sorted_qualified = sorted(qualified, reverse=True)
for rank, count in enumerate(sorted_qualified, start=1):
    print(f"第{rank}名：{count}次")

# 用同一份資料，門檻 100 呼叫
_, qualified_100 = select_clicks(raw_clicks, min_clicks=100)
print(f"門檻100合格結果：{qualified_100}")
print(f"門檻100合格筆數：{len(qualified_100)}")
print(f"門檻100合格合計：{sum(qualified_100)}")
if len(qualified_100) == 0:
    print("無符合門檻的素材")
```
